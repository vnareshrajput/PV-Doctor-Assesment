```python
from pathlib import Path

pr_folder = Path("data/PR")

pr_files = sorted(pr_folder.rglob("*.csv"))

print(len(pr_files))

print(pr_files[:])
```

```python
import pandas as pd

sample_file = pr_files[0]

df = pd.read_csv(sample_file)

print(df)
```

```python
pr_data = []

for file in pr_files:
    df = pd.read_csv(file)

    pr_data.append(df)

    pr_df = pd.concat(
    pr_data,
    ignore_index=True
)
```

## GHI

```python
from pathlib import Path

ghi_folder = Path("data/GHI")

ghi_files = sorted(ghi_folder.rglob("*.csv"))

print(len(ghi_files))

print(ghi_files[:])
```

```python
ghi_data = []

for file in ghi_files:
    df = pd.read_csv(file)
    ghi_data.append(df)

ghi_df = pd.concat(
    ghi_data,
    ignore_index=True
)
```

```python
ghi_df
```

```python
merged_df = pd.merge(
    pr_df,
    ghi_df,
    on="Date",
    how="inner"
)
```

```python
print(type(merged_df))
print(merged_df)
```

```python
merged_df.info()
```

```python
merged_df.shape
```

```python
merged_df["Date"] = pd.to_datetime(merged_df["Date"])
```

```python
merged_df.shape
```

```python
merged_df.info()
```

```python
merged_df = merged_df.sort_values("Date")
merged_df = merged_df.reset_index(drop=True)
```

```python
merged_df.to_csv("data/preprocessed_data.csv", index=False)
```
